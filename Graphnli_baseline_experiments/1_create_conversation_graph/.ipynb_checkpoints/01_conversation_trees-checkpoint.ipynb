# 01 · Conversation trees for GraphNLI

Builds one reply tree per conversation observed on an instance, in the format of
the original GraphNLI code:

```
{"nodes": {toot_id: {"text": ..., "toxicity": ..., "old_conv_ids": {instance_id: conversation_id}}},
 "edges": {child_toot_id: parent_toot_id}}
```

Inputs (written by the DeTox-Fed notebooks in
`DeTox-Fed_experiments/1_creating_conversation_graph`):
`<dataset>_data_with_toxicity.csv` and `replies.csv`.

Output: `<dataset>_graphnli_conversations.pkl`, a stream of pickled records
`{"instance_toot_found_id", "conversation_id", "graph"}` written one at a time, so
that millions of conversations never need to be held in memory together.

````python
import pickle
import time
from pathlib import Path

import networkx as nx
import pandas as pd
````

## Configuration

````python
DATASET = "pleroma"  # or "mastodon"

DATA_DIR = Path("../../data") / DATASET
OUTPUT_DIR = DATA_DIR / "graphnli"

TOOTS_CSV = DATA_DIR / f"{DATASET}_data_with_toxicity.csv"
REPLIES_CSV = DATA_DIR / "replies.csv"
OUTPUT_PKL = OUTPUT_DIR / f"{DATASET}_graphnli_conversations.pkl"

# Smallest conversation kept (number of toots on the instance). The training
# script applies its own, stricter, --min_conversation_length filter.
MIN_TOOTS = {"pleroma": 10, "mastodon": 5}[DATASET]

# Drop conversations containing a toot that the same instance also assigns to
# another conversation (needed for the Pleroma crawl).
DROP_AMBIGUOUS_CONVERSATIONS = {"pleroma": True, "mastodon": False}[DATASET]

# Keep only conversations that are also in the DeTox-Fed conversation table,
# so that both models are evaluated on the same conversations.
RESTRICT_TO_DETOXFED_CONVERSATIONS = False
DETOXFED_CONVERSATIONS_CSV = DATA_DIR / f"final_{DATASET}_nodes_conv_tox.csv"

CONVERSATION_KEY = ["instance_toot_found_id", "conversation_id"]
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
````

## Load toots and reply edges

````python
toots = pd.read_csv(TOOTS_CSV, usecols=CONVERSATION_KEY + ["toot_id", "text", "toxicity"], low_memory=False)
replies = pd.read_csv(REPLIES_CSV)

for df, cols in [(toots, CONVERSATION_KEY + ["toot_id"]), (replies, CONVERSATION_KEY + ["toot_id_parent", "toot_id_child"])]:
    for c in cols:
        df[c] = df[c].astype(str).str.strip()

print(f"Toot observations: {len(toots):,}")
print(f"Conversations:     {toots[CONVERSATION_KEY].drop_duplicates().shape[0]:,}")
print(f"Reply edges:       {len(replies):,}")
````

## Remove ambiguous conversations

A toot that an instance assigns to more than one conversation would appear in two
trees of the same client. Those conversations are removed.

````python
per_instance = toots.groupby(["instance_toot_found_id", "toot_id"])["conversation_id"].nunique()
ambiguous_toots = per_instance[per_instance > 1].reset_index()[["instance_toot_found_id", "toot_id"]]
affected = (
    toots.merge(ambiguous_toots, on=["instance_toot_found_id", "toot_id"], how="inner")[CONVERSATION_KEY]
    .drop_duplicates()
)
print(f"Toots in more than one conversation of the same instance: {len(ambiguous_toots):,}")
print(f"Affected conversations: {len(affected):,}")

if DROP_AMBIGUOUS_CONVERSATIONS and len(affected):
    toots = toots.merge(affected.assign(_drop=1), on=CONVERSATION_KEY, how="left")
    toots = toots[toots["_drop"].isna()].drop(columns="_drop")
    print(f"Toot observations after removal: {len(toots):,}")
````

## Optionally keep only the DeTox-Fed conversations

````python
if RESTRICT_TO_DETOXFED_CONVERSATIONS:
    detoxfed_keys = pd.read_csv(DETOXFED_CONVERSATIONS_CSV, usecols=CONVERSATION_KEY).astype(str).drop_duplicates()
    toots = toots.merge(detoxfed_keys, on=CONVERSATION_KEY, how="inner")
    print(f"Conversations shared with DeTox-Fed: {toots[CONVERSATION_KEY].drop_duplicates().shape[0]:,}")
````

## Keep conversations with at least `MIN_TOOTS` toots

````python
sizes = toots.groupby(CONVERSATION_KEY)["toot_id"].transform("nunique")
toots = toots[sizes >= MIN_TOOTS].copy()
replies = replies.merge(toots[CONVERSATION_KEY].drop_duplicates(), on=CONVERSATION_KEY, how="inner")

print(f"Conversations: {toots[CONVERSATION_KEY].drop_duplicates().shape[0]:,}")
print(f"Reply edges:   {len(replies):,}")
````

## Build and stream the trees

Nodes are the toots of the conversation on that instance; each reply edge points
from the child toot to its parent. Edges whose endpoints are not both in the
conversation are ignored.

````python
def build_conversation_graph(conv_df, rel=None):
    inst = str(conv_df["instance_toot_found_id"].iloc[0])
    conv = str(conv_df["conversation_id"].iloc[0])

    G = nx.DiGraph(instance_toot_found_id=inst, conversation_id=conv)
    toot_ids = set(conv_df["toot_id"])

    for row in conv_df.itertuples(index=False):
        G.add_node(row.toot_id, text=row.text, toxicity=row.toxicity)

    if rel is not None:
        for row in rel.itertuples(index=False):
            if row.toot_id_parent in toot_ids and row.toot_id_child in toot_ids:
                G.add_edge(row.toot_id_child, row.toot_id_parent)
    return G


def to_graphnli_format(G):
    inst = G.graph["instance_toot_found_id"]
    conv = G.graph["conversation_id"]
    return {
        "nodes": {
            str(node): {"text": attrs.get("text", ""), "toxicity": attrs.get("toxicity"), "old_conv_ids": {inst: conv}}
            for node, attrs in G.nodes(data=True)
        },
        "edges": {str(child): str(parent) for child, parent in G.edges()},
    }
````

````python
reply_groups = {key: group for key, group in replies.groupby(CONVERSATION_KEY, sort=False)}
conversations = toots.groupby(CONVERSATION_KEY, sort=False)
total = conversations.ngroups

start = time.time()
n_nodes = n_edges = 0

with open(OUTPUT_PKL, "wb") as f:
    for i, ((inst, conv), conv_df) in enumerate(conversations, start=1):
        G = build_conversation_graph(conv_df, reply_groups.get((inst, conv)))
        n_nodes += G.number_of_nodes()
        n_edges += G.number_of_edges()
        pickle.dump(
            {"instance_toot_found_id": str(inst), "conversation_id": str(conv), "graph": to_graphnli_format(G)},
            f,
            protocol=pickle.HIGHEST_PROTOCOL,
        )

        if i % 10_000 == 0 or i == total:
            elapsed = time.time() - start
            print(f"{i:,}/{total:,} conversations | nodes={n_nodes:,} edges={n_edges:,} | {elapsed / 60:.1f} min", flush=True)

print(f"Saved → {OUTPUT_PKL}")
````

## Check a few trees

````python
def read_stream(path):
    with open(path, "rb") as f:
        while True:
            try:
                yield pickle.load(f)
            except EOFError:
                return


def tree_summary(record):
    data = record["graph"]
    G = nx.DiGraph(list(data["edges"].items()))
    G.add_nodes_from(data["nodes"])
    roots = [n for n in G.nodes if G.out_degree(n) == 0]
    return {
        "instance_toot_found_id": record["instance_toot_found_id"],
        "conversation_id": record["conversation_id"],
        "nodes": G.number_of_nodes(),
        "edges": G.number_of_edges(),
        "roots": len(roots),
        "is_tree": len(roots) == 1 and G.number_of_edges() == G.number_of_nodes() - 1 and nx.is_directed_acyclic_graph(G),
    }


stream = read_stream(OUTPUT_PKL)
pd.DataFrame(tree_summary(next(stream)) for _ in range(min(10, total)))
````
